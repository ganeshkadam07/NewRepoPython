```python
def fibonacci(n):
    a,b=0,1
    for i in range(1,n+1):
        print(a,end=' ')
        a,b=b,a+b

fibonacci(10)
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n % 2 ==0:
            return False
    return True

print(is_prime(10))
```

```python
def count_char(s):
    count={}
    for i in s:
        if i!=' ':
            count[i]=count.get(i,0)+1
    for i,j in count.items():
        print(i,':',j)

print(count_char('Ganesh'))
```

```python
def is_palindrome(n):
    return n==n[::-1]

print(is_palindrome('Ganesh'))
```

```python
def count_vovels(s):
    vovels='aeiouAEIOU'
    return sum(1 for i in s if i in vovels)

print(count_vovels('Ganesh'))
```

```python
def leap_year(year):
    return year % 4==0 and (year %100!=0 or year % 400==0)

print(leap_year(2024))
```

```python
def common_ele(l1,l2):
    return list(set(l1) & set(l2))

print(common_ele([1,2,3],[2,3,4]))
```

```python
def return_word(s):
    return max(len(word) for word in s.split())

print(return_word('Ganesh Purchase on Amazon'))
```

```python
# ✅ 3. Find the First Non-Repeating Character

from collections import Counter

def first_non_repeating(s):
    freq = Counter(s)
    for char in s:
        if freq[char] == 1:
            return char
    return None

# Test
print(first_non_repeating("aabbccde"))  # Output: "d"
```

```python
def reverse_str(data):
    result=''
    for i in data:
        result= i + result
    return result

reverse_str(data='Ganesh')
```
